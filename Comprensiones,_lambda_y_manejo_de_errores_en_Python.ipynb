## 2. Funciones `lambda` y funciones de orden superior

Una **lambda** es una función anónima y conciso. Útil cuando la lógica es breve.


```python
# Lambda básica
suma = lambda a, b: a + b
suma(3, 4)
```

```python
# sorted -> ordena de mayor a menor
# map -> transforma listas o arrays mediante una funcion y devuelve la lista alterada
# filter -> selecciona
# reduce ->

# Usos comunes: sorted, map, filter
palabras = ["perro", "gato", "ratón", "caballo"]

# Ordenar por longitud
ordenadas_longitud = sorted(palabras, key=lambda w: len(w))

# Transformar a mayúsculas
mayus = list(map(lambda w: w.upper(), palabras))

# Filtrar las que contienen 'a'
con_a = list(filter(lambda w: 'a' in w, palabras))

ordenadas_longitud, mayus, con_a
```

## 3. Manejo de errores (`try`, `except`, `else`, `finally`, `raise`)

Estructura típica:
- `try`: código que **puede fallar**.
- `except`: **captura** una o varias excepciones específicas.
- `else`: corre **solo si no hubo excepción**.
- `finally`: corre **siempre** (para liberar recursos, cerrar archivos, etc.).


```python
def dividir(a: float, b:float) -> float:
  if b==0:
    raise ValueError("El denominador no puede ser cero.")
  return a/b
try:
  r = dividir(10,0)
except ValueError as e:
  print("Error:", e)
else:
  print('Resultado:',r)
finally:
  print('Fin del bloque de division.')
```

```python
def division(a: float, b: float) -> float:
  if b==0:
    raise ValueError("El denominador no puede ser cero.")
    return a/b
try:
  r= division(10,2)
except ValueError as e:
  print('Error:',e)
else:
  print('Resultado',r)
finally:
  print('Fin del bloque division')
```

### `try`/`except`/`else`/`finally` en lectura de archivo

```python
nombre = "archivo_inexistente.txt"
try:
    f = open(nombre, "r", encoding="utf-8")
    try:
        contenido = f.read()
    except UnicodeDecodeError:
        print("Error de decodificación.")
    else:
        print("Leí el archivo (simulado).")
    finally:
        f.close()
except FileNotFoundError:
    print(f"No se encontró {nombre!r}.")
```

```python
def media_segura(nums):
    # 1. Validación de lista vacía (ValueError)
    if len(nums) == 0:
        raise ValueError("La lista está vacía")

    # 2. Validación de tipos (TypeError)
    # Ponemos el código peligroso dentro del try
    try:
        total = 0
        for n in nums:
            total += n  # Aquí saltará TypeError si n es un string o None

        media = total / len(nums) # Aquí saltaría ZeroDivisionError
        return float(media)

    except TypeError:
        # Esto atrapa el error si intentaste sumar un string o None
        raise TypeError("La lista contiene valores no numéricos")

    except ZeroDivisionError:
        # Doble protección por si la lista está vacía
        raise ValueError("La lista está vacía")

try:
    resultado = media_segura([1, "hola", 3])
    print(f"La media es: {resultado}")
except TypeError as e:
    print(f"⚠️  Aviso: {e}")
except ValueError as e:
    print(f"⚠️  Aviso: {e}")

print("El programa continúa felizmente...")
```

## 7. Expresiones generadoras y buenas prácticas

### Expresiones generadoras
Parecidas a las comprensiones, pero **producen** valores bajo demanda (lazy). Útiles para **streams** o listas muy grandes.


```python
# Generador: produce cuadrados infinitos (cuidado al iterar)
def cuadrados():
    n = 0
    while True:
        yield n*n
        n += 1

# Tomar los primeros 5
gen = cuadrados()
primeros_5 = [next(gen) for _ in range(5)]
primeros_5
```

### `sum` con generador vs lista

```python
import sys

# usando lista (ocupa memoria)
lista = [x for x in range(10_000)]
mem_lista = sys.getsizeof(lista)

# usando generador (no materializa la lista)
mem_gen = sys.getsizeof((x for x in range(10_000)))

s1 = sum([x for x in range(10_000)])
s2 = sum(x for x in range(10_000))

mem_lista, mem_gen, s1 == s2
```

### Buenas prácticas y errores comunes
- Prefiere comprensiones cuando mejoran **claridad** y concisión.
- Evita comprensiones **anidadas complejas** → usa bucles con nombre intermedio.
- `lambda` para **funciones pequeñas**; si crece, conviértelo en `def` con docstring.
- Captura **excepciones específicas** (no uses `except Exception:` sin motivo).
- Usa `else` cuando quieras ejecutar código **solo si no falló** el bloque `try`.
- `finally` para **limpieza** (archivos, conexiones, etc.).
